```python
def appendWords(word1, word2):
    list1 = list(word1)
    list2 = list(word2)
    result = []

    for i in range(max(len(list1), len(list2))):
        if i < len(list1):
            result.append(list1[i])
        if i < len(list2):
            result.append(list2[i])

    return "".join(result)


print(appendWords("Hello", "World"))  



def MergeWords(word1, word2):
    return word1 + word2

print(MergeWords("Hello", "World"))  







    
```

```python
import math

def gcdOfStrings(str1 , str2):
    if str1+str2 != str2+str1:
        return "Not a GCD !!"
    
    gcd_len = math.gcd(len(str1) , len(str2))
    return str1[:gcd_len]



print(gcdOfStrings("ABCABC", "ABC"))     
print("******************************")
print(gcdOfStrings("ABABAB", "ABAB"))   
print("******************************") 
print(gcdOfStrings("python", "C++")) 
```

```python
def kidsWithCandies(candies, extraCandies):
        maxCandies = max(candies) 
        result = []


        for i in candies :
            if i +extraCandies >= maxCandies:
                result.append(True)

            else:
                 result.append(False)  


        return result



print(kidsWithCandies([2, 3, 5, 1, 3], 3))  

                 
```

Another Solution (Easy)

```python
def kidsWithCandies(candies, extraCandies):
    maxCandies = max(candies)
    return [(i + extraCandies >= maxCandies) for i in candies]

print(kidsWithCandies([2, 3, 5, 1, 3], 3))

#Using Booleans
```

```python
class Node:
    def __init__(self, data):
        self.data = data
        self.next = None


head = Node(10)
head.next = Node(20)
head.next.next = Node(30)


def search(node, value):
    while node:
        if node.data == value:
            return True
        node = node.next
    return False


print(search(head, 20))  
print(search(head, 40))  
```

```python
def reverseVowels(s):
    vowels = set("aeiouAEIOU")
    s = list(s)

    def swap(i, j):
        s[i], s[j] = s[j], s[i]

    i, j = 0, len(s) - 1
    while i < j:
        if s[i] not in vowels:
            i += 1
        elif s[j] not in vowels:
            j -= 1
        else:
            swap(i, j)
            i += 1
            j -= 1

    return "".join(s)


print(reverseVowels("IceCreAm"))

```
